## 1. Module Overview & Foundational Concepts

### 1.1 Definition and Scope of NLP

Natural Language Processing (NLP) is an interdisciplinary field situated at the intersection of **Computer Science**, **Artificial Intelligence (AI)**, and **Computational Linguistics**.

* **Language**: A medium of structured human communication.
* **Natural Language**: A language that evolved organically through human culture and communication without a predefined formal grammar (e.g., English, Hindi, Bengali). Formal grammar rules were formulated *after* the language had already naturally developed.
* **Artificial / Programming Language**: A synthetic language designed with a strict, predefined formal grammar prior to implementation (e.g., Python, C++).

| Dimension | Natural Languages | Artificial / Programming Languages |
| --- | --- | --- |
| **Evolution** | Organic and empirical (Grammar formalized post-evolution) | Designed synthetically (Grammar specified pre-implementation) |
| **Ambiguity** | Highly ambiguous across multiple linguistic levels | Deterministic and unambiguous |
| **Flexibility** | High tolerance for errors, slang, and context shift | Strict syntax; compiler/interpreter throws errors on syntax deviation |

### 1.2 Diversity and Linguistic Scale

Processing natural language computationally is non-trivial due to massive regional and global linguistic diversity:

* **Global Scale**: Over **6,000+ official languages** worldwide, alongside an uncounted number of regional dialects.
* **Indian Subcontinent**: Official census reports record **19,000+ languages and dialects** across India. Regional variations within a single language (e.g., Bengali across West Bengal and Bangladesh) show extreme lexical and phonetic divergences between geographic zones (e.g., Eastern vs. Western dialects of Bangladesh).

## 2. Key Historical Foundations & Theoretical Frameworks

### 2.1 Alan Turing & The Turing Test

* **Alan Turing** is widely regarded as the foundational figure of both Computer Science and NLP.
* **Key Literature**: *"Computing Machinery and Intelligence"* (1950).
* **The Turing Test (Imitation Game)**:
* **Setup**: Three entities placed in isolated partitions—a Machine ($A$), a Human ($B$), and a Human Interrogator/Evaluator ($C$).
* **Objective**: Evaluator $C$ communicates with $A$ and $B$ purely via natural language text/audio channels.
* **Evaluation Criterion**: If Evaluator $C$ cannot reliably distinguish whether responses originate from Machine $A$ or Human $B$, Machine $A$ is declared to have passed the Turing Test.



### 2.2 Noam Chomsky & Formal Linguistics

* **Noam Chomsky** is considered the father of modern formal and computational linguistics.
* **Universal Grammar**: Postulates an innate, underlying structural framework shared across human natural languages.
* **Chomsky Normal Form (CNF)**: A formal context-free grammar specification widely applied in syntactic parsing algorithms.

## 3. Linguistic Ambiguity & Computational Challenges

Ambiguity is the primary bottleneck in natural language understanding. Text can be ambiguous at multiple structural layers.

### 3.1 Structural & Syntactic Ambiguity

Occurs when a single sequence of words can yield multiple valid grammatical parse trees.

* **Example**: *"I saw a girl with a telescope."*
* **Interpretation 1 (VP-attachment)**: The speaker used a telescope as an instrument to observe the girl.
* **Interpretation 2 (NP-attachment)**: The girl was holding/possessing a telescope.


* **Semantic Disambiguation via Context**:
* *"I saw a girl with a bicycle."* $\rightarrow$ Unambiguous: Bicycle attaches to NP (girl has the bicycle).
* *"I saw a bus with a telescope."* $\rightarrow$ Unambiguous: Telescope attaches to VP (instrument used by speaker).



### 3.2 Semantic Ambiguity

Occurs when a word or phrase carries multiple distinct dictionary meanings.

* **Example**: *"Mary had a little lamb."*
* **Sense 1**: Mary owned/kept a small animal (lamb).
* **Sense 2**: Mary consumed/ate a portion of lamb meat.



### 3.3 Pragmatic & Contextual Ambiguity

Occurs when the surface/literal meaning differs significantly from the intended functional meaning.

* **Idioms & Metaphors**: *"Virat Kohli was on fire last night, he totally destroyed the other team."*
* *Literal Surface Meaning*: Kohli was physically combusting.
* *Pragmatic Meaning*: Kohli played exceptionally well.


* **Prepositional Role Ambiguity**:
* *"I ate rice with a spoon."* $\rightarrow$ **Instrument**
* *"I ate rice with curd."* $\rightarrow$ **Auxiliary Food**
* *"I ate rice with Rahul."* $\rightarrow$ **Companionship**



### 3.4 Punctuation and Orthographic Ambiguity

Punctuation drastically alters structural interpretation and semantic boundaries.

* **Example 1**:
* *"Let's eat grandma."* (Implies eating the person)
* *"Let's eat, grandma."* (Direct address to grandma)


* **Example 2**:
* *"A woman without her man is nothing."*
* *"A woman: without her, man is nothing."* (Reverses semantic polarity)



### 3.5 Case Study: The 8-Word "Buffalo" Sentence

> **Sentence**: *"Buffalo buffalo Buffalo buffalo buffalo buffalo Buffalo buffalo"*

This is a syntactically valid English sentence based on three distinct meanings of the word **buffalo**:

1. **Proper Noun / Adjective**: The city of Buffalo, New York ($B_N$).
2. **Noun**: The animal bison ($b_A$).
3. **Verb**: To bully, baffle, or intimidate ($b_V$).

#### Structural Decomposition:

$$\underbrace{\text{Buffalo}_N\text{ buffalo}_A}_{\text{NY Bison (Subject Set 1)}} \quad \underbrace{\underbrace{\text{Buffalo}_N\text{ buffalo}_A}_{\text{NY Bison (Subject Set 2)}} \quad \underbrace{\text{buffalo}_V}_{\text{bully}}}_{\text{Relative Clause (that NY Bison bully)}} \quad \underbrace{\text{buffalo}_V}_{\text{bully (Main Verb)}} \quad \underbrace{\text{Buffalo}_N\text{ buffalo}_A}_{\text{NY Bison (Object Set)}}$$

* **Paraphrase**: *"Bison from Buffalo, NY (whom other bison from Buffalo bully), themselves bully bison from Buffalo."*

```
                 Sentence (S)
                      |
    +-----------------+-----------------+
    |                                   |
  NP (Subject)                       VP (Predicate)
    |                                   |
 +--+--------------+               +----+----+
 |                 |               |         |
NP (NY Bison 1)   Sbar (Rel. Clause) Verb    NP (NY Bison 3)
                   |             (bully)     |
              +----+----+                 +--+--+
              |         |                 |     |
             NP        VP                N     N
        (NY Bison 2)  (bully)

```

### 3.6 Text Processing Bottlenecks

1. **Non-standard Text / Slang**: Social media terminology (e.g., `LOL`, `OMG`).
2. **Segmentation Issues**: Space delimiters fail for compound entities (e.g., *"The New York-New Haven railway"*).
3. **Multi-Word Expressions (MWEs) & Idioms**: Non-compositional meanings where individual word meanings do not sum to the total meaning (e.g., *"hot dog"* refers to food, not a warm canine).
4. **Dynamic Vocabulary / Out-of-Vocabulary (OOV)**: Rapidly evolving neologisms not present in standard dictionaries (e.g., *"unfriend"*, *"retweet"*).
5. **World Knowledge Requirements**: Ambiguities requiring external ground-truth facts (e.g., historical parser errors attributing Michelle Obama as the spouse of George Bush instead of Barack Obama).
6. **Complex Named Entity Recognition (NER)**: Titles containing common words (e.g., *"Let It Be"* as a song title vs. the individual verb tokens *let*, *it*, *be*).

## 4. The Linguistic Processing Hierarchy & NLP Trinity

### 4.1 NLU vs. NLG

* **Natural Language Understanding (NLU)**: Takes raw natural language text and maps it to structured internal representations (e.g., POS tagging, Parsing, Sentiment Analysis, Intent Extraction).
* **Natural Language Generation (NLG)**: Takes structured representations or prompt states and generates fluent natural language sequences (e.g., Machine Translation, Dialogue Systems/Chatbots, Summarization).

### 4.2 The Layered Linguistic Hierarchy

NLP models process text across increasing abstraction levels—from low-level surface morphology up to high-level societal/contextual pragmatics.

```
       [ Pragmatics & Discourse ]    ^  Increasing Model
                   |                 |  Complexity & Connection
             [ Semantics ]           |  to World Knowledge
                   |                 |
              [ Syntax ]             |
                   |                 |
            [ Morphology ]           |  Surface Form Processing

```

### 4.3 The NLP Trinity Paradigm

The core architecture of classical and modern computational linguistics can be represented as a 3D coordinate space:

* **X-Axis (Languages)**: English, Hindi, Bengali, Marathi, French, etc.
* **Y-Axis (Tasks)**: Morphological Analysis, POS Tagging, Chunking, Parsing, Semantic Role Labeling, Coreference Resolution, Generation.
* **Z-Axis (Algorithms / Models)**: Hidden Markov Models (HMM), Maximum Entropy Markov Models (MEMM), Conditional Random Fields (CRF), Deep Neural Networks (DNN), Large Language Models (LLM).

### 5.1 Layer 1: Morphology & Tokenization

Morphology is the study of internal word structure and the formation of words from smaller units called **morphemes** (roots, stems, prefixes, suffixes, infixes, circumfixes).

#### Tokenization Challenges

Simple white-space splitting fails in production systems due to:

* **Numerical Expressions**: `1,000.50` (commas and periods serve as numeric formatters, not delimiters).
* **URLs & Web Entities**: `[https://domain.com/path](https://domain.com/path)` (slashes and dots cannot be naively split).
* **Hashtags & Subwords**: `#greatestdayever` (requires multi-word compound segmentation).

#### Stemming vs. Lemmatization

| Attribute | Stemming | Lemmatization |
| --- | --- | --- |
| **Mechanism** | Heuristic, rule-based truncation of affixes (e.g., Porter/Lancaster stemmers). | Morphological analysis utilizing vocabulary dictionaries and POS tags. |
| **Output String** | May **not** be a valid dictionary word (e.g., `studying` $\rightarrow$ `studi`). | Guarantees a valid canonical base form (**lemma**) (e.g., `studying` $\rightarrow$ `study`). |
| **Speed / Memory** | Extremely fast; minimal computational/memory overhead. | Slower; requires memory lookups and contextual tag analysis. |
| **Use Case** | Fast search index building, basic text normalization. | Fine-grained semantic pipelines, machine translation, NLU. |

#### Morphological Richness

* **Morphologically Poor Languages**: English, Chinese (minimal inflectional variance per word root).
* **Morphologically Rich Languages**: Hindi, Hungarian, Finnish (words carry extensive inflections for tense, gender, number, case, and honorifics).
* *Example in Hindi*: Verbs inflect dynamically based on gender and formality:
* *"I will go"* $\rightarrow$ *मैं जाऊँगा* (Male) / *मैं जाऊँगी* (Female).

### 5.2 Layer 2: Syntactic Analysis

Syntax concerns the formal rules governing how individual words combine into valid phrases and sentences.

#### A. Part-of-Speech (POS) Tagging

Assigning grammatical category labels (e.g., Noun, Verb, Adjective) to each token in a sequence.

* **Standard Tagsets**:
* **Penn Treebank Tagset**: 45 distinct tags (e.g., `PRP` = Personal Pronoun, `VBD` = Verb Past Tense, `DT` = Determiner, `NN` = Singular Noun).
* **C7 Tagset**: 146 fine-grained tags.


* **Contextual POS Ambiguity**:
* *"He went to the **park**."* $\rightarrow$ `park` = `NN` (Noun)
* *"They went to **park** the car."* $\rightarrow$ `park` = `VB` (Verb)



#### B. Chunking (Shallow Parsing)

Grouping adjacent, non-overlapping tokens into syntactically correlated phrases (Noun Phrase `NP`, Verb Phrase `VP`, Prepositional Phrase `PP`).

* **Example Parsing Sequence**:

$$\text{[Mumbai green lights women icons]}_{\text{NP}} \quad \text{[on]}_{\text{PP}} \quad \text{[traffic signals]}_{\text{NP}} \quad \text{[earns]}_{\text{VP}} \quad \text{[global praise]}_{\text{NP}}$$



#### C. Full Parsing (Constituency Parsing)

Constructing a complete structural hierarchy (Parse Tree) representing sentence syntax.

```
                  Sentence (S)
                       |
        +--------------+--------------+
        |                             |
    Noun Phrase (NP)           Verb Phrase (VP)
        |                             |
   Pronoun (PRP)            +---------+---------+
        |                   |                   |
       He               Verb (VBZ)       Noun Phrase (NP)
                            |                   |
                          eats            +-----+-----+
                                          |           |
                                   Determiner (DT) Noun (NN)
                                          |           |
                                          a         mango

```

### 5.3 Layer 3: Semantic Analysis

Semantics transitions NLP from structural manipulation to meaningful extraction.

#### Paradigm 1: Decompositional Semantics

Decomposes lexical meaning into a fixed vector of binary boolean features.

* **Feature Vector Schema**: $[\text{Human}, \text{Female}, \text{Adult}]$
* $\text{Boy} = [+1, -1, -1]^T$
* $\text{Girl} = [+1, +1, -1]^T$
* $\text{Man} = [+1, -1, +1]^T$
* $\text{Woman} = [+1, +1, +1]^T$



#### Paradigm 2: Ontological Semantics

Uses explicitly curated hierarchical knowledge structures and lexical databases (e.g., **WordNet**, Princeton University, ~1995).

* **Primary Lexical Relations**:
* **Synonymy**: Identity of meaning (e.g., *car* $\equiv$ *automobile*).
* **Antonymy**: Opposite of meaning (e.g., *hot* $\leftrightarrow$ *cold*).
* **Hyponymy / Hypernymy**: Is-A subclass/superclass relationship (e.g., *tractor* is a hyponym of *vehicle*).
* **Meronymy / Holonymy**: Part-Of / Whole-Of relationship (e.g., *wheel* is a meronym of *car*).


* **Limitations**: Highly expensive human curation, static nature, fails to keep up with dynamic modern vocabulary shifts.

#### Paradigm 3: Distributional Semantics

The foundation for modern neural embeddings (e.g., Word2Vec). Based on **Firth’s Structural Hypothesis (1957)**:

> *"You shall know a word by the company it keeps."*

#### Construction of Co-occurrence Matrix $M$:

1. Define a sliding context window size $k$ (e.g., $k = 10$ tokens).
2. For a unique vocabulary size $V$, construct a $V \times V$ co-occurrence matrix $M$, where cell $M_{i,j}$ records the total count of times word $j$ appears within context window $k$ of word $i$.
3. Each row vector $\mathbf{v}_i \in \mathbb{R}^V$ acts as a dense or sparse vector representation of target word $i$.

```
Co-occurrence Matrix M (Window size k):
               pet   owner   run   bark   road
  dog   [       5      4      3      2      0   ]  --> Row vector v_dog
  cat   [       6      3      2      0      0   ]  --> Row vector v_cat
  car   [       0      1      0      0      5   ]  --> Row vector v_car

```

#### Vector Similarity Measurement:

To compute the semantic similarity between two word vectors $\mathbf{u}$ and $\mathbf{v}$, use the **Cosine Similarity**:

$$\text{Cosine Similarity}(\mathbf{u}, \mathbf{v}) = \cos(\theta) = \frac{\mathbf{u} \cdot \mathbf{v}}{\Vert{}\mathbf{u}\Vert{}_2 \Vert{}\mathbf{v}\Vert{}_2} = \frac{\sum_{m=1}^{V} u_m v_m}{\sqrt{\sum_{m=1}^{V} u_m^2} \sqrt{\sum_{m=1}^{V} v_m^2}}$$

* **Range**: $[-1, 1]$. A value closer to $1$ signifies high semantic distributional similarity.

### 5.4 Layer 4: Pragmatics & Discourse

Pragmatics analyzes the **negotiation of meaning between speaker and listener** based on explicit and implicit real-world contexts.

* **Conversational Context Example 1**:
* *Query*: *"Do you know what time it is?"*
* *Literal Surface Response*: *"Yes."*
* *Pragmatic Correct Response*: *"It is 2:30 PM."* (Or acknowledging lateness).


* **Conversational Context Example 2**:
* *Query*: *"Can you pass the water bottle?"*
* *Literal Surface Response*: *"Yes, I can."* (Without moving).
* *Pragmatic Action*: Executing the physical act of handing over the bottle.

## 7. Exam-Oriented Practice Questions

### Question 1 (Numerical Vector Calculation)

Given a vocabulary space $V = [\text{pet}, \text{owner}, \text{run}, \text{bark}]$, the extracted co-occurrence vectors for two words $\mathbf{w}_1 = \text{"dog"}$ and $\mathbf{w}_2 = \text{"cat"}$ are:

* $\mathbf{v}_{\text{dog}} = [5, 2, 0, 1]^T$
* $\mathbf{v}_{\text{cat}} = [4, 3, 1, 0]^T$

Compute the exact **Cosine Similarity** between $\mathbf{v}_{\text{dog}}$ and $\mathbf{v}_{\text{cat}}$.

#### Step-by-Step Solution:

1. **Compute Dot Product ($\mathbf{v}_{\text{dog}} \cdot \mathbf{v}_{\text{cat}}$)**:

$$\mathbf{v}_{\text{dog}} \cdot \mathbf{v}_{\text{cat}} = (5 \times 4) + (2 \times 3) + (0 \times 1) + (1 \times 0) = 20 + 6 + 0 + 0 = 26$$


2. **Compute $L_2$ Norm $\Vert{}\mathbf{v}_{\text{dog}}\Vert{}_2$**:

$$\Vert{}\mathbf{v}_{\text{dog}}\Vert{}_2 = \sqrt{5^2 + 2^2 + 0^2 + 1^2} = \sqrt{25 + 4 + 0 + 1} = \sqrt{30}$$


3. **Compute $L_2$ Norm $\Vert{}\mathbf{v}_{\text{cat}}\Vert{}_2$**:

$$\Vert{}\mathbf{v}_{\text{cat}}\Vert{}_2 = \sqrt{4^2 + 3^2 + 1^2 + 0^2} = \sqrt{16 + 9 + 1 + 0} = \sqrt{26}$$


4. **Calculate Cosine Similarity**:

$$\text{Cosine Similarity} = \frac{26}{\sqrt{30} \times \sqrt{26}} = \frac{\sqrt{26}}{\sqrt{30}} = \sqrt{\frac{13}{15}} \approx \mathbf{0.9309}$$



* **Conclusion**: High similarity score ($\approx 0.931$) indicates strong semantic distribution overlap.

### Question 2 (Conceptual MCQ)

**Which of the following statements is TRUE regarding Stemming and Lemmatization?**

* A) Stemming always returns a valid dictionary word, whereas Lemmatization does not.
* B) Lemmatization relies on morphological dictionaries and POS context, guaranteeing a valid canonical base word.
* C) Stemming is computationally slower than Lemmatization because it builds full syntax trees.
* D) WordNet is an algorithmic framework used exclusively for stemming.

**Answer**: **B**

* **Explanation**: Lemmatization uses morphological lexicons and POS context to return a valid base word (lemma). Stemming uses heuristic suffix stripping and may return non-dictionary strings.

### Question 3 (Syntactic Ambiguity Identification)

In the sentence *"I ate rice with a spoon"*, what syntactic function does the prepositional phrase *"with a spoon"* fulfill, and how does it differ semantically from *"with Rahul"* in *"I ate rice with Rahul"*?

#### Solution / Explanation:

* In *"with a spoon"*, the prepositional phrase attaches to the **Verb Phrase (VP)** as an **Instrumental Modifier** (the instrument used to perform the eating action).
* In *"with Rahul"*, the prepositional phrase attaches as a **Companionship / Accompaniment Modifier** (identifying a co-agent).
* Both sentences share an identical surface syntactic structure, proving that syntax alone cannot resolve semantic role labeling without contextual/world knowledge.

## 8. Concise Revision Summary

* **NLP Definition**: Lies at the intersection of **Computer Science, AI, and Computational Linguistics**.
* **Natural vs. Formal**: Natural languages evolved organically without formal initial specifications, making them inherently ambiguous across multiple layers.
* **Key Historical Figures**:
* **Alan Turing**: Introduced the Imitation Game (**Turing Test**, 1950) to evaluate machine intelligence.
* **Noam Chomsky**: Modern linguistics pioneer (**Universal Grammar**, **CNF**).


* **Linguistic Layers (Bottom-Up)**:
1. **Morphology**: Internal word structures, morphemes, FSA-based splitting.
2. **Syntax**: Formal grouping rules (**POS Tagging**, **Chunking**, **Parsing**).
3. **Semantics**: Literal meaning derivation (**Decompositional**, **Ontological/WordNet**, **Distributional/Co-occurrence Matrix**).
4. **Pragmatics & Discourse**: Meaning negotiation based on context and world knowledge.


* **Stemming vs. Lemmatization**: Stemming uses fast, non-dictionary suffix removal; Lemmatization uses POS tags and lexicons to yield valid base lemmas.
* **Distributional Semantics Core Formula**: $\mathbf{w}_i$ is represented by context vector $\mathbf{v}_i$; semantic closeness is computed via **Cosine Similarity**:

$$\frac{\mathbf{u} \cdot \mathbf{v}}{\Vert{}\mathbf{u}\Vert{}_2 \Vert{}\mathbf{v}\Vert{}_2}$$
