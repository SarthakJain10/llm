# Introduction to Large Language Models (LLMs)

This module serves as the foundational introduction to Large Language Models (LLMs), outlining the evolution, core mathematical formulations, and architectural paradigms that define modern Generative AI for text.

## 1. Core Foundations of Language Models (LMs)

A language model is fundamentally a **probability distribution over a sequence of tokens**. A token can be a word, a character, a subword, or even a sequence of numbers.

The primary task of an LM is to compute the probability of a given sequence of tokens. This probability score reflects both the syntactic (grammatical) correctness and the semantic (world knowledge) validity of the sequence.

### Types vs. Tokens

Before computing probabilities, an LM requires a **vocabulary**.

* **Corpus (plural: Corpora):** The large body of text data used for training.
* **Tokens:** The individual instances of words (or subwords/characters) as they appear in the text.
* **Types:** The *unique* tokens collected from the corpus. The vocabulary is strictly a set of these unique types.

### Syntax, Semantics, and Probabilities

A robust language model implicitly learns world knowledge by assigning probabilities based on validity:

* **Valid Syntax & Semantics:** "The man eats the cheese" $\rightarrow$ High probability (e.g., $0.8$).
* **Valid Syntax, Invalid Semantics:** "The cheese eats the man" $\rightarrow$ Low probability (e.g., $0.1$). It has a valid subject-verb-object structure, but violates logical world knowledge.
* **Invalid Syntax & Semantics:** "The man eats cheese the" $\rightarrow$ Extremely low probability (e.g., $0.001$).

### Mathematical Formulation

Given a sequence of tokens $X = (x_1, x_2, \dots, x_L)$, the LM measures the joint probability of the sequence. Using the chain rule of probability, this joint probability is decomposed into a series of conditional probabilities:

$$P(x_1, x_2, \dots, x_L) = P(x_1) \times P(x_2 \vert{} x_1) \times P(x_3 \vert{} x_1, x_2) \dots P(x_L \vert{} x_1, \dots, x_{L-1})$$

### Autoregressive Generation

The conditional probability formulation turns the LM into a **generative model**. To generate text, the model acts as an **autoregressive model**, meaning it uses its own previous outputs as context for future predictions.

For example, given the context $x_1 = \text{"The"}$ and $x_2 = \text{"monsoon"}$, the model evaluates the conditional probability distribution for $x_3$:

* $P(\text{"rains"} \vert{} \text{"The", "monsoon"})$
* $P(\text{"arrived"} \vert{} \text{"The", "monsoon"})$

The model samples from this distribution, selecting a highly probable token (like "rains") as $x_3$, appends it to the sequence, and repeats the process to predict $x_4$.

## 2. Evolution of Architectures: RNNs to Transformers

### The Pre-Transformer Bottleneck

Early sequential models like **RNNs (Recurrent Neural Networks)**, **LSTMs (Long Short-Term Memory)**, and **GRUs (Gated Recurrent Units)** process data sequentially (token by token).

* **The Problem:** The sequential nature prevents parallel processing of the input sequence, creating a major computational bottleneck.
* **Historical Note:** LSTMs/GRUs were initially introduced (around 1997) with gating mechanisms specifically to solve the *vanishing gradient problem* found in standard RNNs.

### The Transformer Breakthrough (2017)

Introduced by Google Brain (Ashish Vaswani et al.), the **Transformer** model revolutionized NLP by processing all input tokens in parallel.

* **Positional Encoding:** Because the Transformer processes inputs in parallel rather than sequentially, it natively lacks awareness of token order. Positional encodings are added to the input embeddings to explicitly provide the model with the sequential position of each token.
* **Architecture:** The original Transformer contained both an **Encoder block** and a **Decoder block**, designed primarily for sequence-to-sequence tasks like machine translation.

## 3. The Three Pillars of LLM Architecture

Following the Transformer, the NLP community split into three primary architectural approaches, which are heavily tested in conceptual exams:

1. **Encoder-Only Models (e.g., BERT)**
* **Introduced:** 2018 (Google).
* **Mechanism:** Uses only the Encoder block of the Transformer. Employs a **Masked Language Modeling (MLM)** training protocol (predicting masked words from bidirectional context).
* **Variants:** RoBERTa (an extensively trained, optimized BERT), DistilBERT, MobileBERT (parameter-efficient smaller versions).
* **Best For:** Classification, sentiment analysis, extractive question answering.


2. **Decoder-Only Models (e.g., GPT Series, LLaMA)**
* **Introduced:** 2018 (OpenAI - GPT-1).
* **Mechanism:** Uses only the Decoder block. Trained strictly as an autoregressive language model (predicting the next token).
* **Best For:** Open-ended text generation, conversational AI.


3. **Encoder-Decoder Models (e.g., T5, BART)**
* **Introduced:** 2019 (Google - T5).
* **Mechanism:** Retains both blocks of the original Transformer. Unifies all NLP tasks into a **"text-to-text"** framework (e.g., inputting "translate English to German: Hello" and outputting "Hallo").
* **Training Strategy:** Span corruption (masking sequences of tokens rather than single words).

## 4. Scaling Laws and Emergent Properties

As models scaled from millions to trillions of parameters, they exhibited **emergent properties**—capabilities they were not explicitly trained for.

* **Zero-Shot Learning (Observed in GPT-2, 1.5B parameters):** The model can perform downstream tasks (like translation or summarization) during inference without seeing any task-specific training examples.
* **In-Context Learning (Observed in GPT-3, 175B parameters):** By providing a few examples of a task within the input prompt (the context), the model adapts and performs the task successfully *without any gradient updates* to its internal neural network parameters.

## 5. Parameter-Efficient Fine-Tuning (PEFT) and Alignment

Training 100+ billion parameter models from scratch is computationally infeasible for most practitioners. The course outlines advanced methods to adapt these models:

* **Knowledge Distillation:** Transferring knowledge from a massive, complex model (Teacher) to a smaller, faster model (Student) so the student mimics the teacher's performance.
* **Adapter-based methods (PEFT):** Freezing the pre-trained model and adding a small number of trainable parameters to adapt the model to specific tasks.
* **Human Alignment:** Techniques like **RLHF** (Reinforcement Learning from Human Feedback) and **DPO** (Direct Preference Optimization) are used to ensure the model's outputs are helpful, harmless, and aligned with human values.

## NPTEL Practice Questions

**Q1. In the context of language modeling, what does the term "types" refer to?**
a) The total number of words in a document.
b) The unique tokens present in a corpus vocabulary.
c) The syntactic categories (nouns, verbs) in a sentence.
d) The specific sequence of characters generated by the model.

**Answer:** b
*Explanation:* While tokens are the individual instances of words in a text, "types" strictly refer to the unique tokens that make up the model's vocabulary.

**Q2. Why is Positional Encoding strictly necessary in a Transformer model, but not in an LSTM?**
a) LSTMs cannot handle sequences longer than 512 tokens.
b) Transformers use gating mechanisms that erase sequence history.
c) Transformers process the entire sequence of input tokens in parallel, bypassing inherent sequential order.
d) Positional encoding solves the vanishing gradient problem found in LSTMs.

**Answer:** c
*Explanation:* RNNs/LSTMs process tokens sequentially, inherently capturing order. Transformers process all inputs simultaneously (in parallel) to avoid computational bottlenecks, meaning order information must be artificially injected via positional encodings.

**Q3. Which of the following best describes the core architectural difference between BERT and the GPT series?**
a) BERT is a decoder-only model, while GPT is an encoder-only model.
b) BERT is an encoder-decoder model, while GPT is an encoder-only model.
c) BERT focuses solely on the encoder block using masked language modeling, while GPT relies solely on the decoder block for autoregressive generation.
d) Both use the exact same architecture, but BERT is trained on cross-lingual data.

**Answer:** c
*Explanation:* BERT utilizes the bidirectional encoder block of the Transformer, while the GPT series is built entirely on the autoregressive decoder block.

**Q4. A user provides a language model with a prompt containing three examples of English-to-French translation, and then asks it to translate a fourth sentence. The model successfully translates it without updating any of its neural network weights. This emergent property is known as:**
a) Parameter-Efficient Fine-Tuning (PEFT)
b) Zero-Shot Learning
c) Span Corruption
d) In-Context Learning

**Answer:** d
*Explanation:* In-context learning (popularized by GPT-3) allows the model to "learn" a task dynamically during inference based purely on the context provided in the prompt, with absolutely no gradient updates.

## Quick Revision Summary

Language models calculate the joint probability of token sequences via the chain rule of conditional probabilities, generating text autoregressively. While early sequence modeling relied on LSTMs to solve vanishing gradients, the 2017 Transformer introduced parallel processing via positional encodings. This led to three distinct lineages: Encoder-only (BERT, used for bidirectional understanding), Decoder-only (GPT/LLaMA, used for generation), and Encoder-Decoder (T5, unifying tasks as text-to-text). As models scaled exponentially in parameter count and training corpora, they unlocked emergent capabilities like zero-shot capability (GPT-2) and in-context learning (GPT-3), paving the way for modern fine-tuning (PEFT, Distillation) and alignment techniques (RLHF, DPO).
