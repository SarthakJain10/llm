## 1. Core Issues in LLM Behavior

When evaluating the behavior of models like ChatGPT, Claude, DeepSeek, or LLaMA, researchers are primarily concerned with two major pitfalls:

* **Inherent Bias:** The model generates varying or discriminatory responses based on sensitive attributes (e.g., country, gender, race) rather than facts. For example, affirming the social acceptance of unmarried cohabitation for one country while denying it for another based on learned stereotypes.
* **Hallucination:** The model fabricates facts, sources, or citations instead of abstaining from answering when it lacks knowledge.
* *Example:* Falsely claiming Charles Lindbergh was the first person on the moon (instead of Neil Armstrong) or generating fake titles of research papers that do not exist.

## 2. The Four Axes of Responsible LLMs

Responsible AI development evaluates models across four primary dimensions:

| Axis | Definition | Key Concepts & Applications |
| --- | --- | --- |
| **Explainability** | Understanding how inputs are processed and attributing outputs to specific inner workings (interpretability). | **Inference Time Intervention:** Using interpretability techniques to intervene in model internals during testing/inference to correct biased or factually incorrect responses. |
| **Fairness** | Identifying the root causes of inherent bias (usually in training data), quantifying it via metrics, and mitigating it. | Seeks to ensure outputs do not reflect stereotypical beliefs or discriminate against specific demographics. |
| **Robustness** | The model's stability against prompt variations and its resiliency against unethical requests. | **Resiliency:** Refraining from answering harmful queries (e.g., substituting "how to make a cake" with "how to make a bomb"). Enforced via **guardrails**. |
| **Safety** | The model's ability to withstand deliberate adversarial attacks designed to break its guardrails. | Evaluates resilience against **Prompt Injection** (black-box attacks altering the prompt) and **White-box attacks** (altering internal representations). |

## 3. Types of Biases in LLMs

Bias is a distortion from an ethical or non-stereotypical response. Biased responses can fall into a direct **stereotype** or a **reverse stereotype**. If left unmitigated, biased LLMs can cause real-world harm, including societal polarization and geopolitical tension.

### A. Racial and Ethnic Bias

Models often absorb systemic biases from their training data. A study analyzing traffic police body-cam transcripts found models learned disparate interaction tones:

* **Black Americans:** Abrupt, rude, and presumptive tone (e.g., starting with first names, assuming guilt).
* **White Americans:** Polite, apologetic tone (e.g., "Sorry to stop you").
Language models exposed to this data can amplify these discriminatory speech patterns.

### B. Political and Entity-Association Bias

Using famous names in a prompt heavily skews the model's completion behavior:

* Names like "Hillary" or "Donald" yield highly polarized (strongly positive or negative/objectionable) text.
* Rare or neutral names (e.g., "Anwey") yield neutral completions because the model lacks deeply ingrained contextual baggage for them.

### C. Religious Bias

A study in *Nature* tested GPT-3 with analogy prompts: `"audacious is to boldness as [Religion] is to [Predict]"`.

* **Muslim:** The model associated the term with violent tokens **66%** of the time.
* **Christian:** The model associated the term with generous or positive tokens (e.g., "faithfulness").

### D. Sentence and Discourse-Level Bias (StereoSet Dataset)

**StereoSet** measures bias across four domains: Gender, Profession, Race, and Religion. It uses fill-in-the-blank tasks: *"Girls tend to be more [BLANK] than boys."*
The model is given three options:

1. **Stereotypical** (e.g., soft)
2. **Anti-stereotype** (reverse stereotype)
3. **Unrelated / Meaningless**

**Metrics to Quantify Bias:**

* **LMS (Language Modeling Score):** The percentage of instances the model chooses a meaningful option (1 or 2) over the meaningless option (3). An ideal model has an LMS of **100**.
* **SS (Stereotype Score):** The percentage of times the model selects the stereotypical option over the anti-stereotypical one. An ideal model has an SS of **50** (perfectly balanced).
* **ICAT Score:** The overarching metric combining LMS and SS. An ideal ICAT score is **100**.

$$\text{ICAT} = \text{LMS} \times \frac{\min(\text{SS}, 100 - \text{SS})}{50}$$

*Findings:* **RoBERTa-base** proved the most balanced, while GPT and BERT families exhibited heavy stereotypical behavior.

### E. Gender-Profession Bias

Tested using **coreference resolution** tasks with ambiguous pronouns.

* *Prompt:* "The doctor phoned the nurse because she was late."
* *Ambiguity:* "She" could refer to either the doctor or the nurse.
* *Result:* **95% of the time**, LLMs associated "she" with the nurse, blindly relying on the stereotype that nursing is a female profession instead of asking clarifying questions.

## 4. Sources of Bias

1. **Training Data Skewness:** LLMs are trained heavily on web data (e.g., Wikipedia). English Wikipedia is heavily skewed toward media, politics, and warfare—topics inherently laden with bias. Balanced data selection is critical to offset this.
2. **Temporal Bias:** Word meanings evolve over time. Training on older texts causes misalignment with modern contexts.
* *Unfriend:* Historically meant "enemy"; now means "removing a social media connection."
* *Degree:* Historically meant "social status hierarchy"; now refers to "academic achievement."
* *Mouse:* Animal/rodent vs. computer peripheral.


3. **Cultural and Linguistic Bias:** There is a vicious cycle favoring **High-Resource Languages** (e.g., English) over **Low-Resource Languages** (e.g., Manipuri, Swahili, endangered languages). High-resource languages get annotated and improved exponentially, while low-resource languages stagnate, posing immense challenges for equitable multilingual LLMs.

## 5. Mitigating Bias: Adversarial Triggers & Evaluation

Mitigation broadly falls into two categories: **Adversarial Triggers** and **In-Context Learning (ICL)** (the latter to be covered in subsequent lectures).

### Adversarial Triggers

Adversarial triggers are specific strings of text prepended to an input prompt to artificially neutralize the model's biased output.

* *Unmitigated Example:* "What was Shanice known for?" -> *LLM:* "Fighting people." (Stereotyped toward Black names).
* *Mitigated Example:* `[Adversarial Trigger] + "What was Shanice known for?"` -> *LLM:* "She is a professor at MIT." (Forces a neutral/positive output).

### Quantifying Bias in Output: Sentiment vs. Regard

To evaluate if a mitigation technique works, we must measure the output's bias.

* **Sentiment Score:** Evaluates the polarity of the *entire sentence*.
* **Regard Score:** Evaluates the polarity *toward a specific demographic or entity*.

> **Example:** *"Person X successfully sustained a livelihood as a beggar for 15 years."*
> * **Sentiment:** Positive/Neutral (due to "successfully sustained").
> * **Regard:** Negative (associating Person X with a "beggar", which is historically held in low social regard).
> *Implementation:* Researchers fine-tune BERT models on hand-curated datasets to predict "Regard" instead of generic sentiment.

## Practice Questions

**Q1. An LLM is asked to complete a paragraph but instead outputs a fake research paper title and non-existent authors. Which issue does this represent?**
A) Inherent Bias
B) Prompt Injection
C) Hallucination
D) Inference Time Intervention
*Answer: C*
*Explanation:* Fabricating facts, sources, or citations when the model doesn't know the answer is defined as hallucination.

**Q2. An attacker modifies a prompt to intentionally bypass an LLM's safety guardrails, tricking it into providing instructions on building a weapon without accessing the model's internal weights. What type of attack is this?**
A) White-box attack
B) Black-box Prompt Injection
C) Coreference Resolution
D) Temporal Bias
*Answer: B*
*Explanation:* A prompt injection is a black-box attack where the user manipulated inputs to break guardrails without having access to the model's internals.

**Q3. Calculate the ICAT score for an LLM that chooses a meaningful option in the StereoSet dataset 90% of the time, and selects the stereotypical option 80% of the time.**
A) 36
B) 72
C) 90
D) 45
*Answer: A*
*Explanation:* $\text{LMS} = 90$. $\text{SS} = 80$. The anti-stereotype percentage is $100 - 80 = 20$.
$\text{ICAT} = 90 \times (\min(80, 20) / 50)$
$\text{ICAT} = 90 \times (20 / 50) = 90 \times 0.4 = 36$.

**Q4. When analyzing the sentence, "He happily survived for a decade as a common thief," why might a researcher prefer the "Regard" metric over the "Sentiment" metric to measure model bias?**
A) Regard analyzes the timeline of the training data.
B) Sentiment might score the sentence as positive due to "happily survived," while Regard correctly identifies the negative social association with "thief" for the entity.
C) Sentiment only works for low-resource languages.
D) Regard calculates the grammatical correctness of the sentence.
*Answer: B*
*Explanation:* Sentiment evaluates the whole sentence's emotional tone, which can be fooled by positive verbs. Regard measures the polarity of societal perception specifically associated with the entity.

**Q5. Why do older training datasets introduce "temporal bias" into modern LLMs?**
A) Because older datasets inherently contain prompt injection attacks.
B) Because words like "degree" or "unfriend" had fundamentally different meanings in the past, leading to modern contextual misinterpretation.
C) Because older texts belong to low-resource languages.
D) Because they artificially inflate the Language Modeling Score (LMS).
*Answer: B*
*Explanation:* Temporal bias occurs when the model learns archaic definitions (e.g., degree as social status) that clash with modern usages (degree as academic achievement).

## Revision Summary

* **Responsible LLMs** focus on Explainability (interpretability), Fairness (mitigating bias), Robustness (resiliency to prompt variations), and Safety (defense against adversarial attacks).
* **Bias manifests** in racial disparities, entity polarization, religious stereotyping, and gender-profession assumptions (often failing coreference resolution by not asking clarifying questions).
* **StereoSet Evaluation:** Measures LMS (ability to pick meaningful text) and SS (balance of stereotyping). Combined mathematically into the **ICAT Score** (Ideal = 100).
* **Sources of Bias:** Skewed web data (too much politics/warfare), temporal drift in definitions, and the dominance of high-resource languages over low-resource ones.
* **Mitigation & Measurement:** Bias can be countered by prepending **Adversarial Triggers**. Output bias is better measured by **Regard** (entity-specific polarity) than general **Sentiment** (overall sentence polarity).
