## Historical Evolution of Neural Networks

Neural networks were initially conceived in the 1940s to mimic human brain interactions using computational approaches.

* **1943-1944:** McCulloch (neuroscientist) and Pitts (logician) proposed the **MPNet** (McCulloch-Pitts Network), the first simple computational model of a neuron.
* **1957-1958:** Frank Rosenblatt proposed the **Perceptron**, considered the first realistic neural network with learnable weights.
* **1965-1968:** The **Multi-Layer Perceptron (MLP)** was proposed to handle more complex decision boundaries.
* **1969:** Minsky and Papert proved that a simple Perceptron could not model the non-linearly separable **XOR function**. This caused a stagnation in AI research known as the **AI Winter** (approx. 1969–1986).
* **1986:** The **Backpropagation** method was popularized, providing a reliable training protocol for multi-layer networks.
* **1989:** The **Universal Approximation Theorem (UAT)** was proposed, stating that an MLP with a single hidden layer can approximate any continuous function to any desired precision.

## Biological to Artificial Neuron Mapping

An artificial neuron mathematically simulates the biological processes of a real neuron:

* **Dendrites (Input):** Receive signals from the environment or other neurons ($x_1, x_2, \dots, x_n$).
* **Soma (Processing):** The central part that aggregates inputs and performs computational processing (Aggregation and Activation).
* **Axon/Synapses (Output):** Transmits the processed signal to the next neuron ($y$).

## Early Neural Models

### 1. The MPNet (McCulloch-Pitts Network)

The MPNet is a basic model handling exclusively binary inputs and outputs. All inputs carry equal importance (no weights).

* **Inputs:** Binary variables $x_i \in \{0, 1\}$.
* **Operation:** Sums the inputs $\sum x_i$. If the sum reaches a threshold $\theta$, the output is $1$; otherwise, $0$.
* **Modeling Boolean Logic (Linearly Separable):**
* **AND Gate:** For $x_1, x_2$, output is $1$ only if both are $1$. Set $\theta = 2$. Line: $x_1 + x_2 \ge 2$.
* **OR Gate:** Output is $1$ if at least one input is $1$. Set $\theta = 1$. Line: $x_1 + x_2 \ge 1$.


* **Limitation:** It cannot model the **XOR** function because the points $(0,0) \to 0, (0,1) \to 1, (1,0) \to 1, (1,1) \to 0$ cannot be separated by a single linear boundary.

### 2. The Perceptron

The Perceptron introduces **weights** to signify the relative importance of different inputs, mimicking how real-world decisions prioritize certain features.

* **Inputs:** Real numbers $x_1, x_2, \dots, x_n$.
* **Parameters:** Weights $w_1, w_2, \dots, w_n$ and a bias term $w_0$.
* **Aggregation:** Computes the weighted sum: $a = \sum_{i=1}^n w_i x_i + w_0$.
* **Activation:** Passes the sum through a modified Signum (step) function $g(x)$:

$$g(x) = \begin{cases} 1, & \text{if } x \ge 0 \\ 0, & \text{if } x < 0 \end{cases}$$


* **Example (AND Gate):** Weights $w_1=1, w_2=1$, bias $w_0 = -1.5$. Equation: $x_1 + x_2 - 1.5 \ge 0$.
* **Limitation:** Like the MPNet, a single Perceptron provides only a linear decision boundary and still **cannot classify XOR**.

## Multi-Layer Perceptron (MLP)

To solve non-linear problems like XOR, MLPs introduce hidden layers and non-linear activation functions. This architecture is a **Feed Forward Network**.

* **Depth:** The total number of layers (often excluding the input layer).
* **Width:** The number of neurons in a specific layer.
* **Notation:**
* $W_{ji}^{(l)}$: Weight connecting the $i$-th neuron in layer $l-1$ to the $j$-th neuron in layer $l$.
* $a_j^{(l)}$: Pre-activation output (aggregation) of neuron $j$ in layer $l$.
* $z_j^{(l)}$: Post-activation output of neuron $j$ in layer $l$.



### Two-Step Neuron Operation

1. **Aggregation ($A$):** Computes the weighted sum of inputs plus bias.

$$a_j^{(l)} = \sum_{i=1}^m W_{ji}^{(l)} z_i^{(l-1)} + W_{j0}^{(l)}$$


2. **Activation ($H$):** Applies a non-linear function to the aggregated sum.

$$z_j^{(l)} = h(a_j^{(l)})$$



### Solving the XOR Problem with MLP

A simple MLP with 2 inputs, 1 hidden layer (2 neurons), and 1 output neuron can perfectly model the XOR gate using modified step activations:

1. **Hidden Neuron 1 (OR logic):** $w_1=1, w_2=1, b=-0.5 \implies h_1 = g(x_1 + x_2 - 0.5)$
2. **Hidden Neuron 2 (AND logic):** $w_1=1, w_2=1, b=-1.5 \implies h_2 = g(x_1 + x_2 - 1.5)$
3. **Output Neuron:** Combines $h_1$ and $h_2$ with weights $1$ and $-1$, bias $-0.5$.

$$y = g(h_1 - h_2 - 0.5)$$



*Verification for $x_1=1, x_2=0$:* $h_1 = g(0.5) = 1$; $h_2 = g(-0.5) = 0$. $y = g(1 - 0 - 0.5) = g(0.5) = 1$. (Correct for XOR).

## Activation Functions

Activation functions inject **non-linearity** into the network, allowing it to approximate complex functions. The derivative of these functions is crucial for backpropagation.

| Function | Formula / Logic | Key Characteristics & Issues |
| --- | --- | --- |
| **Linear** | $g(x) = x$ | Derivative is a constant ($1$). Fails to introduce non-linearity. |
| **Sigmoid** | $\sigma(x) = \frac{1}{1 + e^{-x}}$ | Squashes input to $(0, 1)$. Derivative is $\sigma(x)(1 - \sigma(x))$. **Issues:** Causes Vanishing Gradient (derivative nears 0 for $\vert{}x\vert{} > 3$); not symmetric around 0. |
| **Tanh** | $\tanh(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}}$ | Squashes input to $(-1, 1)$. Symmetric around 0 (better optimization). Derivative is $1 - \tanh^2(x)$. **Issue:** Still suffers from Vanishing Gradient. |
| **Softmax** | $\frac{e^{x_i}}{\sum e^{x_j}}$ | Converts an input vector into a probability distribution. |
| **ReLU** | $\max(0, x)$ | Linear for positive values, 0 for negative. Derivative is 1 or 0. **Issue:** Dead Neuron Problem (if pre-activation is negative, neuron permanently outputs 0). |
| **Leaky ReLU** | $x$ if $x \ge 0$, else $0.01x$ | Solves the Dead Neuron Problem by allowing a small gradient for negative values. |
| **PReLU** | $x$ if $x \ge 0$, else $ax$ | Similar to Leaky ReLU, but $a$ is a learnable parameter. |
| **Concat ReLU** | $[\text{ReLU}(x), \text{ReLU}(-x)]$ | Vector of size 2 capturing both positive and negative magnitude features. |
| **ELU** | $x$ if $x \ge 0$, else $\alpha(e^x - 1)$ | Introduces an exponential component for negative values. |
| **GELU** | $x \cdot \Phi(x)$ | Multiplies input by the standard normal cumulative distribution function $\Phi(x)$. Widely used in modern LLMs. |
| **GLU** | $A \otimes \sigma(WA + b)$ | Gated Linear Unit. Uses element-wise multiplication ($\otimes$) to dynamically scale (gate) inputs based on learnable weights. |
| **Swish** | $x \cdot \sigma(\beta x)$ | Can act as linear or ReLU depending on the tunable/learnable parameter $\beta$. |
| **SwiGLU** | $x \otimes \text{Swish}(Wx + b)$ | Combines Swish and GLU mechanisms. Very popular in current architectures. |

## Training via Backpropagation

Backpropagation calculates how much each weight contributed to the network's final error, allowing the network to update weights using **Stochastic Gradient Descent (SGD)**.

**1. Error Function (Square Loss):**


$$E = \frac{1}{2} \sum_{k=1}^K (y_k - t_k)^2$$


Where $y_k$ is the predicted output and $t_k$ is the ground truth. (The $\frac{1}{2}$ is used so it cancels out cleanly when differentiated).

**2. Weight Update Rule:**


$$W_{new} = W_{old} - \eta \frac{\partial E}{\partial W_{old}}$$


Where $\eta$ is the learning rate.

**3. Applying the Chain Rule:**
To update a weight deep in the network (e.g., $W_{ji}^{(1)}$ connecting the input to the first hidden layer), the gradient must cascade backwards through every intermediate operation:


$$\frac{\partial E}{\partial W_{ji}^{(1)}} = \frac{\partial E}{\partial z_j^{(1)}} \cdot \frac{\partial z_j^{(1)}}{\partial a_j^{(1)}} \cdot \frac{\partial a_j^{(1)}}{\partial W_{ji}^{(1)}}$$

* $\frac{\partial a_j^{(1)}}{\partial W_{ji}^{(1)}} = x_i$ (the input data).
* $\frac{\partial z_j^{(1)}}{\partial a_j^{(1)}} = h'(a_j^{(1)})$ (the derivative of the activation function, e.g., $1 - (z_j^{(1)})^2$ for $\tanh$).
* $\frac{\partial E}{\partial z_j^{(1)}}$ sums the errors propagating backward from the subsequent layer's weights.

### The Vanishing Gradient Problem

As the network depth increases, calculating the gradient for early layers requires multiplying many intermediate derivatives. If activation derivatives are small (e.g., $< 1$ in Sigmoid/Tanh), the continuous multiplication shrinks the gradient exponentially toward zero. Consequently, early layers fail to update and learn.

### Regularization and Hyperparameters

To prevent the model from **overfitting** (memorizing the training data instead of generalizing):

* **Early Stopping:** Halt training before the model achieves 0% error on the training set.
* **L1/L2 Regularization:** Add a penalty term to the error function based on the size of the weights.
* **Dropout:** Randomly "freeze" a fraction of neurons during each training pass, forcing the network to distribute learning and making it sparse.
* **Learning Rate Decay:** Dynamically reduce $\eta$ over epochs (e.g., $\eta_{new} = \eta_{old} \cdot (1 - \text{decay} \times \text{epoch})$) to stabilize convergence.

## NPTEL Practice Questions

**Q1: Which of the following is the primary limitation of the McCulloch-Pitts (MPNet) model that led to its decline?**
A) It cannot process real-numbered inputs.
B) It cannot be trained using backpropagation.
C) It cannot model non-linearly separable logic functions like XOR.
D) It suffers from the dead neuron problem.
*Answer:* **C**. Minsky and Papert demonstrated in 1969 that simple single-layer linear models (both MPNet and Perceptron) cannot classify XOR because the data points cannot be separated by a straight line.

**Q2: During backpropagation, a deep neural network utilizing the Sigmoid activation function stops learning in its initial layers. What is the most likely mathematical cause of this?**
A) The Dead Neuron Problem.
B) The Vanishing Gradient Problem caused by the Sigmoid derivative maximizing at 0.25.
C) The learning rate $\eta$ is set too high.
D) The model is heavily overfitting the training data.
*Answer:* **B**. The derivative of Sigmoid is $\sigma(x)(1 - \sigma(x))$. Its maximum value is $0.25$. Multiplying multiple values $\le 0.25$ through the chain rule across layers causes the gradients to exponentially decay to zero.

**Q3: Calculate the output of a Leaky ReLU activation function $f(x)$ given a pre-activation aggregation value of $a = -5.0$.**
A) $0$
B) $-0.05$
C) $-5.0$
D) $0.05$
*Answer:* **B**. The Leaky ReLU function is defined as $x$ if $x \ge 0$, else $0.01x$. Since $-5.0 < 0$, the output is $0.01 \times -5.0 = -0.05$.

## Key Exam Takeaways

* **Chronology:** MPNet (1943) $\to$ Perceptron (1957) $\to$ MLP/XOR limitations identified (1969) $\to$ AI Winter $\to$ Backprop popularized (1986) $\to$ UAT (1989).
* **XOR Solution:** Requires at least an MLP with one hidden layer containing non-linear activations. A single Perceptron operates strictly via linear boundaries.
* **Network Parameters:**
* *Depth:* Number of layers.
* *Width:* Number of neurons per layer.


* **Activation Functions:**
* *Sigmoid/Tanh:* Differentiable but cause vanishing gradients. Tanh is zero-centered; Sigmoid is not.
* *ReLU:* Computationally efficient (derivative is 1 or 0), but susceptible to dead neurons.
* *Modern Variants:* GELU, GLU, Swish, and SwiGLU utilize gating, probability distributions, and parameterization to overcome basic activation limits and are standard in LLMs.


* **Backpropagation:** Built entirely on the mathematical **chain rule**. It updates weights in the opposite direction of the gradient: $W = W - \eta \nabla E$.
* **Overfitting Countermeasures:** Dropout, Early Stopping, L1/L2 Regularization.
